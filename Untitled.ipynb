```python
import pandas as pd
from astropy.table import Table

# Reemplazá el texto entre comillas si tu archivo FITS tiene otro nombre
tab = Table.read("dr7MPA_WISE.fits")
df = tab.to_pandas()

# Esto te muestra las primeras 5 filas para comprobar que cargó bien
df.head()
```

```python
# 1. Veamos la lista completa de las 150 columnas para buscar las magnitudes
print("Todas las columnas del catálogo:")
print(df.columns.tolist())

# 2. Vamos a definir nuestro objetivo (Target)
target_col = "SMASS_MEDIAN"

# 3. Limpiamos las galaxias con masas falsas (los -1 que vimos)
# Nos quedamos solo con las filas donde la masa sea mayor a 0
df_limpio = df[df[target_col] > 0].copy()

print(f"\nGalaxias originales: {len(df)}")
print(f"Galaxias después de limpiar masas falsas: {len(df_limpio)}")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Definimos las columnas
target_col = "SMASS_MEDIAN"
feature_cols = ["MODELU", "MODELG", "MODELR", "MODELI", "MODELZ"]

# 2. Armamos la mesa de trabajo (cambié la forma de seleccionarlo para evitar el error)
work = df_limpio.loc[:, [target_col] + feature_cols].copy()

# 3. Limpieza de infinitos usando el float nativo de Python
work = work.replace([np.inf, -np.inf], float('nan')).dropna()

# 4. Calculamos los colores para darle más información física al modelo
work["u_g"] = work["MODELU"] - work["MODELG"]
work["g_r"] = work["MODELG"] - work["MODELR"]
work["r_i"] = work["MODELR"] - work["MODELI"]
work["i_z"] = work["MODELI"] - work["MODELZ"]

print(f"Galaxias perfectas listas para el modelo: {len(work)}")

# 5. Separamos Entradas (X) y Salidas (y)
X = work.drop(columns=[target_col])
y = work[target_col]

# 6. Separamos en Entrenamiento (80%) y Prueba (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1234
)

print(f"Cantidad de galaxias para entrenar: {len(X_train)}")
print(f"Cantidad de galaxias para el examen (test): {len(X_test)}")
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. El modelo base (Tonto)
print("Entrenando Línea Base (Dummy)...")
base = DummyRegressor(strategy="mean")
base.fit(X_train, y_train)
pred_base = base.predict(X_test)

# 2. El modelo real (Random Forest)
print("Entrenando Random Forest...")
# El manual de tu proyecto recomienda 300 árboles[cite: 184], pero para la primera prueba 
# lo bajamos a 100 así no tenés que esperar horas. Después lo podemos subir.
rf = RandomForestRegressor(n_estimators=100, min_samples_leaf=2, n_jobs=-1, random_state=1234)
rf.fit(X_train, y_train)
pred_rf = rf.predict(X_test)

# 3. La Libreta de Calificaciones (Métricas)
def imprimir_metricas(nombre, y_verdad, y_prediccion):
    rmse = np.sqrt(mean_squared_error(y_verdad, y_prediccion))
    mae = mean_absolute_error(y_verdad, y_prediccion)
    r2 = r2_score(y_verdad, y_prediccion)
    print(f"--- Resultados de {nombre} ---")
    print(f"MAE  (Error Absoluto): {mae:.4f}")
    print(f"RMSE (Error Cuadrático): {rmse:.4f}")
    print(f"R²   (Puntaje de 0 a 1): {r2:.4f}\n")

print("\n--- ¡TERMINÓ EL ENTRENAMIENTO! ---\n")
imprimir_metricas("Línea Base (El tonto)", y_test, pred_base)
imprimir_metricas("Random Forest (Bosque Aleatorio)", y_test, pred_rf)
```

```python
# Evaluamos el modelo en los apuntes que usó para estudiar (Train)
pred_rf_train = rf.predict(X_train)

print("--- Comparación de Sobreajuste (Overfitting) ---")
print("RMSE en el Entrenamiento: {:.4f}".format(np.sqrt(mean_squared_error(y_train, pred_rf_train))))
print("RMSE en el Examen (Test) : {:.4f}".format(np.sqrt(mean_squared_error(y_test, pred_rf))))
```

```python
print("Entrenando Random Forest Controlado...")

# Aumentamos min_samples_leaf y limitamos max_depth para frenar el sobreajuste
rf_controlado = RandomForestRegressor(
    n_estimators=100, 
    min_samples_leaf=10, 
    max_depth=20, 
    n_jobs=-1, 
    random_state=1234
)

# Entrenamos de nuevo
rf_controlado.fit(X_train, y_train)

# Medimos en ambos conjuntos
pred_rf_train_cont = rf_controlado.predict(X_train)
pred_rf_test_cont = rf_controlado.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, pred_rf_train_cont))
rmse_test = np.sqrt(mean_squared_error(y_test, pred_rf_test_cont))

print("\n--- Resultados del Bosque Controlado ---")
print(f"RMSE en el Entrenamiento: {rmse_train:.4f}")
print(f"RMSE en el Examen (Test): {rmse_test:.4f}")
print(f"R² en el Examen         : {r2_score(y_test, pred_rf_test_cont):.4f}")
```

```python
import matplotlib.pyplot as plt

# 1. Extraemos la importancia de las variables (qué tanto las usó el modelo)
importancias = rf_controlado.feature_importances_
nombres_columnas = X_train.columns

# 2. Ordenamos los datos de mayor a menor para que el gráfico quede prolijo
indices_ordenados = importancias.argsort()

# 3. Armamos el gráfico de barras horizontales
plt.figure(figsize=(10, 6))
plt.barh(nombres_columnas[indices_ordenados], importancias[indices_ordenados], color='#2ab0ff')
plt.xlabel("Nivel de Importancia (Fracción de utilidad)")
plt.ylabel("Variables Fotométricas")
plt.title("¿Qué información usa el Bosque para predecir la masa estelar?")
plt.tight_layout()
plt.show()
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import warnings
from sklearn.exceptions import ConvergenceWarning

# Ocultamos advertencias por si la red decide frenar temprano
warnings.filterwarnings("ignore", category=ConvergenceWarning)

print("Armando la tubería y entrenando la Red Neuronal (MLP)...")
print("(Paciencia, las redes son matemáticas puras y llevan su tiempo)")

# 1. Armamos la tubería: Escalador + Red Neuronal [cite: 116, 117, 118]
mlp = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(64, 64), # Dos capas ocultas de 64 "neuronas" cada una [cite: 120]
        activation="relu",
        solver="adam",
        max_iter=500,
        early_stopping=True,
        random_state=1234
    )
)

# 2. Entrenamos [cite: 125]
mlp.fit(X_train, y_train)

# 3. Le tomamos el examen al igual que al bosque [cite: 126]
pred_mlp_train = mlp.predict(X_train)
pred_mlp_test = mlp.predict(X_test)

rmse_mlp_train = np.sqrt(mean_squared_error(y_train, pred_mlp_train))
rmse_mlp_test = np.sqrt(mean_squared_error(y_test, pred_mlp_test))

# 4. Resultados
print("\n--- Resultados de la Red Neuronal (MLP) ---")
print(f"RMSE en el Entrenamiento: {rmse_mlp_train:.4f}")
print(f"RMSE en el Examen (Test): {rmse_mlp_test:.4f}")
print(f"R² en el Examen         : {r2_score(y_test, pred_mlp_test):.4f}")
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, r2_score
import warnings
from sklearn.exceptions import ConvergenceWarning

# Ocultamos advertencias de convergencia
warnings.filterwarnings("ignore", category=ConvergenceWarning)

print("Entrenando la Red Neuronal Mejorada (¡Aprovechá para prepararte unos mates!)...")

# Tubería con hiperparámetros más agresivos
mlp_mejorado = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(128, 64), # Capacidad matemática ampliada
        activation="relu",
        solver="adam",
        max_iter=1000,                # Más tiempo para aprender
        early_stopping=False,         # Apagamos la rendición temprana
        random_state=1234
    )
)

# Entrenamos
mlp_mejorado.fit(X_train, y_train)

# Evaluamos
pred_mlp_train_mej = mlp_mejorado.predict(X_train)
pred_mlp_test_mej = mlp_mejorado.predict(X_test)

rmse_mlp_train_mej = np.sqrt(mean_squared_error(y_train, pred_mlp_train_mej))
rmse_mlp_test_mej = np.sqrt(mean_squared_error(y_test, pred_mlp_test_mej))

# Resultados
print("\n--- Resultados de la Red Neuronal (Mejorada) ---")
print(f"RMSE en el Entrenamiento: {rmse_mlp_train_mej:.4f}")
print(f"RMSE en el Examen (Test): {rmse_mlp_test_mej:.4f}")
print(f"R² en el Examen         : {r2_score(y_test, pred_mlp_test_mej):.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Calculamos los residuos (Predicción menos la Real) usando los datos del examen
residuos_rf = pred_rf_test_cont - y_test

# Preparamos un lienzo con 3 espacios (1 fila, 3 columnas)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# --- GRÁFICO 1: Predicha vs Real ---
# Usamos alpha=0.01 porque al haber +150.000 galaxias, los puntos se superponen mucho.
# Esto crea un efecto de "densidad" (las zonas más oscuras tienen más galaxias).
ax1.scatter(y_test, pred_rf_test_cont, alpha=0.01, s=2, color='royalblue')

# Trazamos la línea de identidad (uno a uno)
min_val = min(y_test.min(), pred_rf_test_cont.min())
max_val = max(y_test.max(), pred_rf_test_cont.max())
ax1.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2, label='Línea 1:1 (Ideal)')

ax1.set_xlabel('Masa Estelar Real (Catálogo)')
ax1.set_ylabel('Masa Estelar Predicha (Modelo)')
ax1.set_title('Masa Predicha frente a Real')
ax1.legend()

# --- GRÁFICO 2: Residuos vs Real ---
ax2.scatter(y_test, residuos_rf, alpha=0.01, s=2, color='seagreen')
# Trazamos la línea del error cero
ax2.axhline(0, color='red', linestyle='--', lw=2)

ax2.set_xlabel('Masa Estelar Real')
ax2.set_ylabel('Residuo (Predicha - Real)')
ax2.set_title('Análisis de Residuos')

# --- GRÁFICO 3: Histograma de Residuos ---
# Un histograma nos muestra si los errores tienen forma de campana de Gauss centrada en 0
ax3.hist(residuos_rf, bins=100, color='purple', alpha=0.7)
ax3.axvline(0, color='red', linestyle='--', lw=2)

ax3.set_xlabel('Valor del Residuo')
ax3.set_ylabel('Frecuencia (Cantidad de Galaxias)')
ax3.set_title('Histograma de Residuos')

# Ajustamos los márgenes para que quede prolijo y mostramos
plt.tight_layout()
plt.show()
```

```python
print("Armando el Conjunto C y entrenando...")

# 1. Copiamos el Conjunto B que ya teníamos armado (Magnitudes + Colores)
X_train_C = X_train.copy()
X_test_C = X_test.copy()

# 2. Le pegamos la columna 'Z' (Redshift) trayéndola del catálogo original 'df'
# Usamos los índices para asegurarnos de que a cada galaxia le toque su Redshift correcto
X_train_C['Z'] = df.loc[X_train.index, 'Z']
X_test_C['Z'] = df.loc[X_test.index, 'Z']

# 3. Entrenamos el Bosque Definitivo con esta nueva súper tabla
rf_C = RandomForestRegressor(
    n_estimators=100, 
    min_samples_leaf=10, 
    max_depth=20, 
    n_jobs=-1, 
    random_state=1234
)
rf_C.fit(X_train_C, y_train)

# 4. Evaluamos
pred_C_test = rf_C.predict(X_test_C)

rmse_C = np.sqrt(mean_squared_error(y_test, pred_C_test))
r2_C = r2_score(y_test, pred_C_test)

print(f"\n--- Resultados del Conjunto C (Con Redshift) ---")
print(f"RMSE en el Examen: {rmse_C:.4f}")
print(f"R² en el Examen  : {r2_C:.4f}")
print(f"(R² anterior del Conjunto B: 0.8801)")
```

```python
print("Entrenando la Red Neuronal con el Conjunto C (Con Redshift)...")
print("(Dale tiempo, que ahora está masticando cosmología pura)")

# Reutilizamos la tubería poderosa pero entrenamos con el Conjunto C
mlp_C = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(128, 64), 
        activation="relu",
        solver="adam",
        max_iter=1000,                
        early_stopping=False,         
        random_state=1234
    )
)

# Entrenamos (Acá transpiran los procesadores)
mlp_C.fit(X_train_C, y_train)

# Evaluamos en el examen final
pred_mlp_C_test = mlp_C.predict(X_test_C)

rmse_mlp_C = np.sqrt(mean_squared_error(y_test, pred_mlp_C_test))
r2_mlp_C = r2_score(y_test, pred_mlp_C_test)

# Resultados
print("\n--- Resultados de la Red Neuronal (Conjunto C) ---")
print(f"RMSE en el Examen: {rmse_mlp_C:.4f}")
print(f"R² en el Examen  : {r2_mlp_C:.4f}")
print(f"(R² anterior de la Red sin Redshift: 0.8666)")
```
